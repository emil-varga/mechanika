```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
ti = 0
tf = 5
a = 1
N = 10

ts = np.linspace(ti, tf, N)
v = a*ts**2
priemerne_rychlosti = 0.5*(v[:-1] + v[1:])
priemerne_casy = 0.5*(ts[:-1] + ts[1:])

fig, [ax_v, ax_s] = plt.subplots(1, 2, sharex=True)

ax_v.plot(ts, v, '-o')
ax_v.stairs(priemerne_rychlosti, edges=ts, fill=True)

s = np.zeros(len(ts))
np.cumulative_sum(priemerne_rychlosti*np.diff(ts), out=s[1:])
ax_s.plot(ts, a*ts**3/3, '--')
ax_s.plot(ts, s)
```
