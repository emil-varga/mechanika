```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def nakresli_trajektorii(predpis, t, ax=None):
    """ Vytvor graf trajektorie podla predpisu, ktory je vyhodnoteny v casoch t """
    if ax is None:
        fig, ax = plt.subplots(1, 2, layout="constrained")

    ax[0].set_aspect("equal")

    x, y = predpis(t)
    ax[0].plot(x, y, label="trajektorie")
    ax[1].plot(t, x, label="x-komponenta")
    ax[1].plot(t, y, label="y-komponenta")

    ax[0].set_xlabel("x")
    ax[0].set_ylabel("y")

    ax[1].legend(loc="upper right")
    ax[1].set_xlabel("cas")
    ax[1].set_ylabel("pozice")

def nakresli_rychlost(predpis, t, ax=None):
    """ Vytvor graf rychlosti trajektorie podla predpisu, ktory je vyhodnoteny v casoch t """
    if ax is None:
        fig, ax = plt.subplots(1, 2, layout="constrained")

    ax[0].set_aspect("equal")

    x, y = predpis(t)
    vx = np.diff(x)/np.diff(t)
    vy = np.diff(y)/np.diff(t)
    ax[0].plot(vx, vy, label="rychlost")
    ax[1].plot(t[:-1], vx, label="x-komponenta")
    ax[1].plot(t[:-1], vy, label="y-komponenta")

    ax[0].set_xlabel("v_x")
    ax[0].set_ylabel("v_y")

    ax[1].legend(loc="upper right")
    ax[1].set_xlabel("cas")
    ax[1].set_ylabel("rychlost")
```

## Kruhovy pohyb

```python
def kruhovy_pohyb(t):
    T = 1
    r = 1
    omega = 2*np.pi/T
    phis = omega*t
    return r*np.cos(phis), r*np.sin(phis)

nakresli_trajektorii(kruhovy_pohyb, np.arange(0, 1, 0.01))
fig, ax = plt.subplots(1,2)
nakresli_rychlost(kruhovy_pohyb, np.arange(0, 1, 0.001), ax)
nakresli_rychlost(kruhovy_pohyb, np.arange(0, 1, 0.2), ax)
```

## Pohyb po zmensujucej sa spiralie

```python
def spirala_pohyb(t):
    T = 1
    r = np.exp(-1*t)
    omega = 2*np.pi/T
    phis = omega*t
    return r*np.cos(phis), r*np.sin(phis)

nakresli_trajektorii(spirala_pohyb, np.arange(0, 3, 0.01))
```

## Superpozice kmitani a rotace

```python
def kmitani_a_rotace_pohyb(t):
    T_rot = 1
    omega_rot = 2*np.pi/T_rot

    T_kmit = 1/9
    omega_kmit = 2*np.pi/T_kmit

    r = 1 + 0.25*np.cos(omega_kmit*t)
    phis = omega_rot*t
    return r*np.cos(phis), r*np.sin(phis)

nakresli_trajektorii(kmitani_a_rotace_pohyb, np.arange(0, 1, 0.01))
```
