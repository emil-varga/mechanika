```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# numericka integracia pohybovej rovnice

def posun(x0, y0, v0x, v0y, g, dt, Cd):
    x = x0
    y = y0
    vx = v0x
    vy = v0y
    x_list = [x0]
    y_list = [y0]
    
    # integrujeme az po bod dopadu (y=0)
    while y >= 0:
        v = np.sqrt(vx**2 + vy**2)

        # vypocitame zrychlenie na zaklade odporu vzduchu a gravitacie
        ax = -Cd * v * vx
        ay = -g - Cd * v * vy
        
        # zmenime rychlost pomocou zrychlenia
        vx += ax * dt
        vy += ay * dt
        
        # zmenime polohu pomocou rychlosti
        x += vx * dt
        y += vy * dt
        
        x_list.append(x)
        y_list.append(y)
    
    return x_list, y_list
```

```python
x0 = 0
y0 = 0
v0x = 10
v0y = 10
g = 9.81
dt = 0.001

for Cd in [0, 0.1, 0.2]:
    x, y = posun(x0, y0, v0x, v0y, g, dt, Cd=Cd)
    plt.plot(x, y, label=f"Cd={Cd}")
plt.legend()
plt.title("Sikmy vrh s odporem vzduchu")
```

```python
x0 = 0
y0 = 0
g = 9.81
dt = 0.001

v0 = 20

for alpha in np.linspace(0, 90, 9):
    v0x = v0 * np.cos(alpha * np.pi / 180)
    v0y = v0 * np.sin(alpha * np.pi / 180)
    x, y = posun(x0, y0, v0x, v0y, g, dt, Cd=0.1)
    plt.plot(x, y, label=f"$\\alpha$={alpha:.2f} deg")
plt.legend()
plt.title("Sikmy vrh s odporem vzduchu")
```
